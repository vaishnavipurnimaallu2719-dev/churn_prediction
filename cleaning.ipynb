```python
# ==========================================
# TELCO CUSTOMER CHURN - DATA CLEANING
# ==========================================

import pandas as pd
import os


# ==========================================
# 1. LOAD RAW DATASET
# ==========================================

# cleaning.ipynb is inside Cleaning/
# Dataset is a sibling folder
df = pd.read_csv("../Dataset/Telco-Customer-Churn.csv")

print("Dataset loaded successfully!")
print("Original Shape:", df.shape)


# ==========================================
# 2. SAVE ORIGINAL DATASET
# ==========================================

df.to_csv(
    "original_dataset.csv",
    index=False
)

print("Original dataset saved.")


# ==========================================
# 3. BASIC DATA INFORMATION
# ==========================================

print("\n========== BEFORE CLEANING ==========")

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# ==========================================
# 4. CLEAN COLUMN NAMES
# ==========================================

df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("\nCleaned Column Names:")
print(df.columns.tolist())


# ==========================================
# 5. SAVE DUPLICATE ROWS
# ==========================================

duplicate_rows = df[df.duplicated(keep=False)]

duplicate_rows.to_csv(
    "duplicate_rows.csv",
    index=False
)

print("\nDuplicate rows saved.")


# ==========================================
# 6. REMOVE DUPLICATE ROWS
# ==========================================

duplicates_removed = df.duplicated().sum()

df = df.drop_duplicates()

print("Duplicates removed:", duplicates_removed)


# ==========================================
# 7. CLEAN CATEGORICAL COLUMNS
# ==========================================

categorical_columns = df.select_dtypes(
    include=["object"]
).columns.tolist()

for column in categorical_columns:

    df[column] = (
        df[column]
        .astype(str)
        .str.strip()
    )

print("\nCategorical values cleaned.")


# ==========================================
# 8. CONVERT TOTAL CHARGES TO NUMERIC
# ==========================================

if "totalcharges" in df.columns:

    df["totalcharges"] = pd.to_numeric(
        df["totalcharges"],
        errors="coerce"
    )

print("TotalCharges converted to numeric.")


# ==========================================
# 9. HANDLE MISSING NUMERICAL VALUES
# ==========================================

numerical_columns = df.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

for column in numerical_columns:

    missing_count = df[column].isnull().sum()

    if missing_count > 0:

        median_value = df[column].median()

        df[column] = df[column].fillna(
            median_value
        )

        print(
            f"{column}: {missing_count} missing values "
            f"filled with median = {median_value}"
        )


# ==========================================
# 10. HANDLE MISSING CATEGORICAL VALUES
# ==========================================

categorical_columns = df.select_dtypes(
    include=["object"]
).columns.tolist()

for column in categorical_columns:

    missing_count = df[column].isnull().sum()

    if missing_count > 0:

        mode_value = df[column].mode()[0]

        df[column] = df[column].fillna(
            mode_value
        )

        print(
            f"{column}: {missing_count} missing values "
            f"filled with mode = {mode_value}"
        )


# ==========================================
# 11. FINAL DATA VALIDATION
# ==========================================

print("\n========== AFTER CLEANING ==========")

print("\nFinal Shape:")
print(df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nRemaining Duplicate Rows:")
print(df.duplicated().sum())

print("\nFinal Data Types:")
print(df.dtypes)


# ==========================================
# 12. CREATE CLEANING SUMMARY
# ==========================================

cleaning_summary = pd.DataFrame({
    "Metric": [
        "Original Rows",
        "Final Rows",
        "Rows Removed",
        "Original Columns",
        "Final Columns",
        "Duplicate Rows Remaining",
        "Missing Values Remaining"
    ],

    "Value": [
        len(pd.read_csv("../Dataset/Telco-Customer-Churn.csv")),
        len(df),
        len(pd.read_csv("../Dataset/Telco-Customer-Churn.csv")) - len(df),
        len(df.columns),
        len(df.columns),
        df.duplicated().sum(),
        df.isnull().sum().sum()
    ]
})

cleaning_summary.to_csv(
    "data_cleaning_summary.csv",
    index=False
)

print("\nCleaning summary saved.")


# ==========================================
# 13. SAVE FINAL CLEANED DATASET
# ==========================================

df.to_csv(
    "cleaned_dataset.csv",
    index=False
)

print("\n==========================================")
print("DATA CLEANING COMPLETED SUCCESSFULLY")
print("==========================================")

print("\nFiles created inside Cleaning folder:")

print("1. original_dataset.csv")
print("2. duplicate_rows.csv")
print("3. data_cleaning_summary.csv")
print("4. cleaned_dataset.csv")
```
