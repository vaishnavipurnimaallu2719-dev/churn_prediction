```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../Cleaning/cleaned_dataset.csv")

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("Churn Distribution:")
print(df["churn"].value_counts())

plt.figure(figsize=(6, 4))

sns.countplot(
    data=df,
    x="churn"
)

plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")

plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.boxplot(
    data=df,
    x="churn",
    y="monthlycharges",
    ax=axes[0]
)

axes[0].set_title("Monthly Charges vs Churn")
axes[0].set_xlabel("Churn")
axes[0].set_ylabel("Monthly Charges")


sns.boxplot(
    data=df,
    x="churn",
    y="tenure",
    ax=axes[1]
)

axes[1].set_title("Tenure vs Churn")
axes[1].set_xlabel("Churn")
axes[1].set_ylabel("Tenure")

plt.tight_layout()
plt.show()
```

```python

import pandas as pd

df = pd.read_csv("../Cleaning/cleaned_dataset.csv")

print("Dataset loaded:", df.shape)

# 1. Check exact duplicate rows
print("Total duplicate rows:", df.duplicated().sum())

# 2. Check duplicate customer IDs
id_columns = [
    col for col in df.columns
    if col.strip().lower().replace("_", "") == "customerid"
]

if id_columns:
    id_col = id_columns[0]

    print("\nCustomer ID column:", id_col)
    print("Duplicate customer IDs:", df[id_col].duplicated().sum())
else:
    print("\nCustomer ID column not found.")

# 3. Check the number of unique values in each column
print("\nUnique values in each column:")
print(df.nunique())

# 4. Check missing values again
print("\nMissing values:")
print(df.isnull().sum())
```

```python
import pandas as pd

# Load cleaned dataset
df = pd.read_csv("../Cleaning/cleaned_dataset.csv")

# 1. Dataset size
print("Dataset Shape:", df.shape)

# 2. Column names
print("\nColumn Names:")
print(df.columns.tolist())

# 3. Missing values in every column
print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

# 4. Duplicate checks
print("\nDuplicate Rows:", df.duplicated().sum())
print("Duplicate Customer IDs:", df["customerid"].duplicated().sum())

# 5. Data types
print("\nData Types:")
print(df.dtypes)

# 6. Target variable distribution
print("\nChurn Distribution:")
print(df["churn"].value_counts())

print("\nChurn Percentage:")
print(
    (df["churn"].value_counts(normalize=True) * 100).round(2)
)

# 7. Summary of numerical columns
print("\nNumerical Summary:")
print(df.describe())

# 8. Summary of categorical columns
print("\nCategorical Summary:")
print(df.describe(include="object").T)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.countplot(
    data=df,
    x="contract",
    hue="churn",
    ax=axes[0]
)

axes[0].set_title("Contract Type vs Churn")
axes[0].set_xlabel("Contract")
axes[0].set_ylabel("Customer Count")
axes[0].tick_params(axis="x", rotation=20)


sns.countplot(
    data=df,
    x="paymentmethod",
    hue="churn",
    ax=axes[1]
)

axes[1].set_title("Payment Method vs Churn")
axes[1].set_xlabel("Payment Method")
axes[1].set_ylabel("Customer Count")
axes[1].tick_params(axis="x", rotation=30)

plt.tight_layout()
plt.show()
```
